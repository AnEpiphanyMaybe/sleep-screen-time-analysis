## Research Question

Does the average daily sleep and the average length of daily smartphone usage of university students have an association with overall quarterly academic performance, measured by average quarter GPA?

## Background and Prior Work

<p>Smartphones have become an almost inseparable part of college student's lives, and consequently, the habit of using them late into the night has become increasingly common. While there is lots of documentation on how poor sleep affects academic performance, less attention has been given to the role late night smartphone usage (between 11PM - 6AM) plays in that process. College students need to balance between late-night entertainment, messaging, and media scrolling with the sleep time they need to reinforce what they learned and to perform well academically. The 11PM threshold is chosen because it represents the boundary after which blue light exposure from screens most aggressively suppresses the production of melatonin and delays sleep, and it aligns with the recommended bedtime range for young adults seeking 7-9 hours of sleep to have healthy cognitive function. <a name="cite_ref-7"></a><a href="#cite_note-7"><sup>7</sup></a></p>

<p>Instead of directly measuring sleep, this project examines whether smartphone usage duration, sleep duration, and sleep consistency together correlate with quarterly GPA among college students. To do this we draw on multiple complementary public datasets: the StudentLife dataset provides objective sensor-based phone activity and sleep data paired with GPA for Dartmouth students; the CMU Sleep dataset provides Fitbit-measured sleep duration and consistency paired with GPA for a larger U.S. sample; the Add Health dataset provides self-reported screen use and sleep alongside transcript GPA for a nationally representative U.S. sample; and the India Screen Time dataset provides daily screen time broken down by usage type alongside GPA across two semesters. Together these datasets allow us to triangulate the relationship between screen time, sleep, and academic performance across different populations and measurement methods. <a name="cite_ref-3"></a><a href="#cite_note-3"><sup>3</sup></a></p>

This dataset comes from the a published academic study called StudentLife that tracked the daily behavior, mental health, and academic performance of college students using smartphone sensor data. The study collected continuous passive data (e.g., phone usage, environmental context) along with survey and academic records for a cohort of undergraduate students at Dartmouth College. Data was collected over an academic term using students’ personal smartphones, allowing for fine-grained observation of behavioral patterns. Academic performance, such as GPA, was obtained from institutional records, making it more reliable than self-reported measures.

The most important metrics in this dataset are as follows. The dark sensor records periods when the phone is in a dark environment, which can serve as a proxy for inactivity or potential sleep. The phonelock sensor captures when the phone is locked or unlocked, providing a basis for estimating phone usage behavior. The EMA (experience sampling) data contains self-reported responses throughout the day, including stress levels, mood, and sleep-related questions, typically recorded on ordinal scales. The education data includes academic outcomes such as GPA on a 0–4 scale, where higher values indicate stronger academic performance. Additional data sources, such as app_usage, conversation, and bluetooth, provide context about social interaction and daily behavior, although these are not directly used in all analyses.

A major concern with this dataset is that many key variables are not directly measured but must be inferred from sensor data. For example, sleep is not explicitly recorded and must be approximated from periods of phone inactivity and darkness, which can introduce measurement error. Similarly, phone usage is inferred from phone lock and unlock events, which may not capture all forms of interaction. The dataset is also subject to missing data, and this missingness may not be random. For instance, students who disable sensors or use their phones less frequently may systematically differ from those with more complete data. Additionally, the sample consists of students from a single university who voluntarily participated in a smartphone sensing study, which introduces self-selection bias and limits generalizability. Finally, while the dataset provides rich behavioral detail, it does not directly measure all aspects of interest (such as exact sleep or full digital engagement), requiring careful interpretation of derived variables.

```python
#Used imports
import json
import numpy as np
import pandas as pd

#StudentLife helper functions
def read_csv_if_exists(path):
    """Read a CSV if it exists, otherwise return None."""
    path = Path(path)
    if path.exists():
        return pd.read_csv(path)
    return None


def to_datetime_seconds(series):
    """Convert Unix seconds to pandas datetime."""
    return pd.to_datetime(pd.to_numeric(series, errors="coerce"), unit="s", errors="coerce")


def get_user_ids():
    """Find users from available phonelock files."""
    files = list((Path(f"{RAW_DIR}/dataset/sensing/phonelock")).glob("phonelock_u*.csv"))
    return sorted(file.stem.replace("phonelock_", "") for file in files)
```

```python
#Creates a interim, daily-user level, aggregated dataset from the multiple data files
all_daily_rows = []
user_ids = get_user_ids()

for user_id in user_ids:
    #Phone usage from phonelock
    lock_path = Path(f"{RAW_DIR}/dataset/sensing/phonelock/phonelock_{user_id}.csv")
    lock = read_csv_if_exists(lock_path)
    if lock is None or lock.empty:
        continue
    lock["start_dt"] = to_datetime_seconds(lock["start"])
    lock["end_dt"] = to_datetime_seconds(lock["end"])
    lock = lock.dropna(subset=["start_dt", "end_dt"])
    lock = lock[lock["end_dt"] > lock["start_dt"]]
    if lock.empty:
        continue
    lock["date"] = lock["start_dt"].dt.date
    lock["locked_hours"] = (lock["end_dt"] - lock["start_dt"]).dt.total_seconds() / 3600
    daily = lock.groupby("date", as_index=False)["locked_hours"].sum()

    #Assuming if the phone is locked, it is not actively being used
    daily["phone_usage_hours"] = (24 - daily["locked_hours"]).clip(lower=0, upper=24)
    daily = daily.drop(columns="locked_hours")
    daily["user_id"] = user_id

    # Sleep inferred from dark sensor
    dark_path = Path(f"{RAW_DIR}/dataset/sensing/dark/dark_{user_id}.csv")
    dark = read_csv_if_exists(dark_path)

    if dark is not None and not dark.empty:
        dark["start_dt"] = to_datetime_seconds(dark["start"])
        dark["end_dt"] = to_datetime_seconds(dark["end"])
        dark = dark.dropna(subset=["start_dt", "end_dt"])
        dark = dark[dark["end_dt"] > dark["start_dt"]]
        dark["sleep_hours"] = (dark["end_dt"] - dark["start_dt"]).dt.total_seconds()/3600

        #Estimated bounds of darkness for realistic sleep
        MIN_SLEEP_HOURS = 0
        MAX_SLEEP_HOURS = 16
        sleep_like = dark[
            (dark["sleep_hours"] >= MIN_SLEEP_HOURS) &
            (dark["sleep_hours"] <= MAX_SLEEP_HOURS)
        ].copy()

        # Assigned sleep to the day the sleep episode starts
        sleep_like["date"] = sleep_like["start_dt"].dt.date

        # Summed all valid sleep_like intervals that start on the same date
        sleep_daily = (
            sleep_like
            .groupby("date", as_index=False)["sleep_hours"]
            .sum()
        )
        sleep_daily["sleep_hours"] = sleep_daily["sleep_hours"].clip(lower=0, upper=16)
        daily = daily.merge(sleep_daily, on="date", how="left")

    else:
        daily["sleep_hours"] = np.nan

    # Stress from EMA responses
    stress_path = Path(f"{RAW_DIR}/dataset/EMA/response/Stress/Stress_{user_id}.json")
    stress_rows = []
    if stress_path.exists():
        with open(stress_path) as f:
            records = json.load(f)
        for record in records:
            if "resp_time" not in record:
                continue
            stress_value = record.get("level")
            if stress_value is None:
                stress_value = record.get("null")
            stress_rows.append({
                "date": pd.to_datetime(record["resp_time"], unit="s").date(),
                "stress_level": pd.to_numeric(stress_value, errors="coerce"),
            })
    if stress_rows:
        stress = pd.DataFrame(stress_rows).dropna(subset=["stress_level"])
        stress_daily = stress.groupby("date", as_index=False)["stress_level"].mean()
        daily = daily.merge(stress_daily, on="date", how="left")
    else:
        daily["stress_level"] = np.nan

    all_daily_rows.append(daily)

daily_behavior = pd.concat(all_daily_rows, ignore_index=True)
daily_behavior = daily_behavior[["user_id", "date", "sleep_hours", "phone_usage_hours", "stress_level"]]

print("Shape:", daily_behavior.shape)
print("Unique users:",daily_behavior["user_id"].nunique())
print("Unique days:",daily_behavior["date"].nunique())
print(daily_behavior.head())
```

```python
#Duplicates and missingness checking
print("Duplicate user-date rows:", daily_behavior.duplicated(["user_id", "date"]).sum())

#Overall missingness
print("Overall missingness:")
print(daily_behavior.isna().mean())

#Missingness by user
print("Missingness by user:")
missing_by_user = (daily_behavior.groupby("user_id")[["sleep_hours","stress_level"]].apply(lambda x: x.isna().mean()))
display(missing_by_user.describe())

#Missingness by date
print("Missingness by date:")
missing_by_date = (daily_behavior.groupby("date")[["sleep_hours","stress_level"]].apply(lambda x: x.isna().mean()))
display(missing_by_date.describe())

#Systematic check
print("Means with missing sleep included or not:")
display(daily_behavior.groupby(daily_behavior["sleep_hours"].isna())[["phone_usage_hours", "stress_level"]].mean())
```

Average missingness of sleep hours is relatively low while missingness of stress levels is very high. The variability of missingness of sleep by user is relatively high, but the variability by day is relatively low. On the other hand, the variability, in stress is high both by user and day showing that it is likely not random, and combined with the high degree of missingness likely will lead to bias. The column is chosen to be dropped for more clear analysis.

```python
#Dropping stress_level
daily_behavior = daily_behavior.drop(columns = ["stress_level"])
daily_behavior["flag_sleep_missing"] = daily_behavior["sleep_hours"].isna().astype(int)
daily_behavior["flag_phone_missing"] = daily_behavior["phone_usage_hours"].isna().astype(int)
daily_behavior["flag_low_sleep_coverage"] = (daily_behavior.groupby("user_id")["sleep_hours"].transform(lambda x: x.notna().mean()) < 0.5).astype(int)

#Save interim dataset
daily_behavior.to_csv(f"{INTERIM_DIR}/daily_behavior.csv", index=False)
print("Saved interim dataset:", f"{INTERIM_DIR}/daily_behavior.csv")
```

```python
#Aggregate user-day level data to user level data
user_behavior = (daily_behavior.groupby("user_id", as_index=False).agg(avg_sleep_hours=("sleep_hours", "mean"),\
    avg_phone_usage_hours=("phone_usage_hours", "mean"), days_observed=("date", "nunique"),percent_flag_sleep_missing=("flag_sleep_missing", "mean"),\
    percent_flag_phone_missing=("flag_phone_missing", "mean"), percent_flag_low_sleep=("flag_low_sleep_coverage", "mean")))

#Combine with academic performance data
grades = read_csv_if_exists(Path(f"{RAW_DIR}/dataset/education/grades.csv"))
grades = grades.rename(columns={"uid": "user_id"})
grades.columns = grades.columns.str.strip().str.lower().str.replace(" ", "_")
user_behavior = user_behavior.merge(grades, on="user_id", how="left")

# Deadlines included as potential confounder for performance, and proxy for amount of coursework
deadlines = read_csv_if_exists(Path(f"{RAW_DIR}/dataset/education/deadlines.csv"))
if deadlines is not None:
    deadlines = deadlines.rename(columns={"uid": "user_id"})
    deadline_long = deadlines.melt(id_vars="user_id", var_name="date", value_name="deadline_count")
    deadline_long["deadline_count"] = pd.to_numeric(deadline_long["deadline_count"], errors="coerce").fillna(0)
    workload = deadline_long.groupby("user_id", as_index=False)["deadline_count"].mean()
    workload = workload.rename(columns={"deadline_count": "avg_deadline_count"})
    user_behavior = user_behavior.merge(workload, on="user_id", how="left")
```

```python
#Remove users without valid GPAs
user_behavior = user_behavior[(user_behavior["gpa_all"].isnull()==False)|(user_behavior["gpa_13s"].isnull()==False)|(user_behavior["cs_65"].isnull()==False)].reset_index()
user_behavior = user_behavior.drop(columns="percent_flag_phone_missing") #Since 0 for all entries
```

```python
#Final status with cleaning
user_behavior = user_behavior.drop(columns=["index"])
print("Shape:", user_behavior.shape)
display(user_behavior)

#Missing values in final dataset
missing = pd.DataFrame({
    "amount_missing": user_behavior.isna().sum(),
    "percent_missing": (user_behavior.isna().mean() * 100).round(1),
})
print(missing[missing["amount_missing"] > 0] if not missing.empty else "Nothing missing")

print("Final outliers")
for col in ["avg_sleep_hours", "avg_phone_usage_hours", "days_observed", "percent_flag_sleep_missing", "gpa_all"]:
    if col in user_behavior.columns:
        print(f"{col}: min={user_behavior[col].min():.2f}, max={user_behavior[col].max():.2f}")

# Save final processed dataset.
user_behavior.to_csv(Path(f"{PROCESSED_DIR}/cleaned_dataset.csv"), index=False)
print("Saved processed dataset:", f"{PROCESSED_DIR}/cleaned_dataset.csv")
```

```python
# Analyze whether there are outliers or suspicious entries in the variables we are interested in.
vars = data2_subset.columns
fig, axes = plt.subplots(1, len(vars), figsize=(18, 4))
for ax, col in zip(axes, vars):
    data2_subset[col].plot(kind='box', ax=ax)
axes = axes.flatten()
plt.tight_layout()
```

```python
# From the plots, we can tell that every variable has several outliers. We will do some calculation to inspect the number of outliers more closely.
def iqr(arr):
    return np.percentile(arr, 75) - np.percentile(arr, 25)
for col in vars:
    column = data2_subset[col]
    num_outliers = (column < column.median() - 1.5 * iqr(column)).sum() + (column > column.median() + 1.5 * iqr(column)).sum()
    prop_outliers = num_outliers / data2_subset.shape[0]
    print(f'{col} has {num_outliers} outliers, accounting for {prop_outliers} of the total number of observations')
```

**From the calculation, we see that sleep_hours has the most number of outliers (over 5%), and is left-skewed. This can be explained by the fact that most students sleep around 7 hours per day, but students who don't get a lot of sleep are very spread out among the different number of hours.** Other than that, the proportions of outliers of the variables in the dataset are within 5%, so there is no suspicious issue about the data.

### CMU Sleep Study: Sleep Duration, Consistency, and GPA

This dataset comes from a published academic study tracking the sleep behavior and academic performance of 634 first-year college students across three U.S. universities: Carnegie Mellon University, University of Washington, and University of Notre Dame. Sleep was recorded using Fitbit wearable devices over approximately one month during the spring academic term, and GPA records were obtained directly from university registrars, making the academic performance measure more reliable than self-reported alternatives.

The most important metrics in this dataset are as follows. `TotalSleepTime` measures average nightly sleep duration in minutes — healthy sleep for young adults falls between 420 and 540 minutes (7–9 hours), and values below 360 minutes (6 hours) have been associated with significant cognitive impairment and lower academic performance. `bedtime_mssd` is the mean successive squared difference of bedtime in minutes squared, measuring night-to-night variability in sleep timing; a higher value indicates more irregular sleep schedules, which is directly relevant to our consistency hypothesis. `midpoint_sleep` is the average midpoint of the sleep episode measured in minutes after 11PM — for example, a value of 240 corresponds to a sleep midpoint of 3AM, indicating a late sleep schedule that overlaps with our 11PM–6AM window of interest. `daytime_sleep` captures average minutes of sleep outside the main nightly episode; elevated values may suggest chronic sleep deprivation being compensated through napping. `term_gpa` and `cum_gpa` both measure academic performance on a 4.0 scale, with `term_gpa` reflecting the specific term during which sleep was tracked and `cum_gpa` reflecting prior academic history.

A major concern with this dataset is that it contains no smartphone screen time data whatsoever, meaning it can only address the sleep → GPA portion of our research question and not the screen time → GPA portion directly. Additionally, the sample is restricted to first-year students, who may behave differently from upperclassmen in terms of sleep habits and academic adjustment. The study also only tracked sleep for roughly one month rather than the full academic quarter, which may not capture the full range of sleep behavior over a term. Finally, participants were drawn from only three universities, limiting generalizability to broader college populations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
# ── 1. Load raw data ──────────────────────────────────────────────────────────
cmu_raw = pd.read_csv('data/00-raw/cmu-sleep.csv')

print("Shape:", cmu_raw.shape)
print("\nColumn names:", cmu_raw.columns.tolist())
print("\nData types:\n", cmu_raw.dtypes)
cmu_raw.head()
```

```python
# ── 2. Demonstrate tidiness ───────────────────────────────────────────────────
# Tidy data: each row = one student (one observation), each column = one variable
# This dataset is already in wide format with one row per student — already tidy.
print("Number of duplicate rows:", cmu_raw.duplicated().sum())
print("\nEach row represents one student — dataset is tidy.")
```

```python
# ── 3. Demonstrate size ───────────────────────────────────────────────────────
print(f"Number of observations (rows): {cmu_raw.shape[0]}")
print(f"Number of variables (columns): {cmu_raw.shape[1]}")
```

```python
# ── 4. Missingness analysis ───────────────────────────────────────────────────
missing = cmu_raw.isnull().sum()
missing_pct = (missing / len(cmu_raw) * 100).round(2)
missing_df = pd.DataFrame({
    'missing_count': missing, 
    'missing_pct': missing_pct
})
print("Missing data summary:")
print(missing_df[missing_df['missing_count'] > 0])

# Visualize missingness
if missing_df['missing_count'].sum() > 0:
    plt.figure(figsize=(10, 4))
    missing_pct[missing_pct > 0].sort_values().plot(kind='barh', color='steelblue')
    plt.xlabel('% Missing')
    plt.title('Missing Data by Column — CMU Sleep')
    plt.tight_layout()
    plt.show()
else:
    print("No missing values found.")
```

```python
# ── 5. Outlier detection ──────────────────────────────────────────────────────
key_vars = ['TotalSleepTime', 'bedtime_mssd', 'midpoint_sleep', 
            'daytime_sleep', 'cum_gpa', 'term_gpa']
key_vars = [v for v in key_vars if v in cmu_raw.columns]

# Boxplots to visually identify outliers
fig, axes = plt.subplots(1, len(key_vars), figsize=(18, 4))
for ax, col in zip(axes, key_vars):
    cmu_raw[col].dropna().plot(kind='box', ax=ax, title=col)
plt.suptitle('Boxplots of Key Variables — CMU Sleep', y=1.02)
plt.tight_layout()
plt.show()

# Flag suspicious entries
# TotalSleepTime: < 60 min likely device error; > 720 min (12 hrs) implausible
if 'TotalSleepTime' in cmu_raw.columns:
    suspicious_sleep = cmu_raw[
        (cmu_raw['TotalSleepTime'] < 60) | (cmu_raw['TotalSleepTime'] > 720)
    ]
    print(f"Suspicious TotalSleepTime entries (< 60 or > 720 min): {len(suspicious_sleep)}")

# GPA: flag values outside valid 0.0–4.0 range
for gpa_col in ['cum_gpa', 'term_gpa']:
    if gpa_col in cmu_raw.columns:
        bad_gpa = cmu_raw[(cmu_raw[gpa_col] < 0) | (cmu_raw[gpa_col] > 4.0)]
        print(f"Out-of-range {gpa_col} entries: {len(bad_gpa)}")
```

```python
# ── 6. Clean the data ─────────────────────────────────────────────────────────
cmu_clean = cmu_raw.copy()

# Drop rows missing any key analysis variables
# Justification: all three variables (sleep duration, sleep consistency, GPA) 
# are required for our analysis — a row missing any of them is unusable.
cmu_clean = cmu_clean.dropna(subset=key_vars, how='any')

# Remove physiologically implausible sleep values
if 'TotalSleepTime' in cmu_clean.columns:
    cmu_clean = cmu_clean[
        (cmu_clean['TotalSleepTime'] >= 60) & (cmu_clean['TotalSleepTime'] <= 720)
    ]

# Remove out-of-range GPA values
for gpa_col in ['cum_gpa', 'term_gpa']:
    if gpa_col in cmu_clean.columns:
        cmu_clean = cmu_clean[
            (cmu_clean[gpa_col] >= 0) & (cmu_clean[gpa_col] <= 4.0)
        ]

print(f"Shape after cleaning: {cmu_clean.shape}")
print(f"Rows dropped: {len(cmu_raw) - len(cmu_clean)}")
```

```python
# ── 7. Save to processed ──────────────────────────────────────────────────────
os.makedirs('data/02-processed', exist_ok=True)
cmu_clean.to_csv('data/02-processed/cmu_sleep_clean.csv', index=False)
print("Saved to data/02-processed/cmu_sleep_clean.csv")
```

## Abstract

College students increasingly use smartphones late into the night, and the consequences of poor sleep on academic performance are well-documented. However, less is known about how sleep duration, smartphone usage, and academic outcomes interact across diverse student populations. This project investigates whether daily sleep duration and smartphone screen time are associated with quarterly GPA among university students, using a cross-dataset approach to assess whether observed patterns generalize across different populations and measurement methods.

We analyzed four complementary datasets: the StudentLife dataset from Dartmouth (n=48), the Global University Students Performance Habits dataset (n=10,000), the Student Screen Time vs CGPA dataset from India (n=550), and the CMU Sleep dataset from CMU/UW/Notre Dame (n=634). For each dataset, we examined distributions of key variables, tested relationships between sleep, screen time, and GPA using permutation tests and linear regression, and controlled for relevant demographic and academic confounders.

Across all four datasets, results pointed in a consistent direction: students who slept more and used their phones less tended to earn higher GPAs, with sleep duration emerging as the most robust predictor. The CMU Sleep dataset showed a statistically significant positive association between sleep duration and GPA (p ≈ 0.004) that held after controlling for gender, race, first-generation status, and units enrolled. Larger datasets like the Global University Students Performance Habits dataset and the India Screen Time dataset corroborated these findings, while the smaller StudentLife dataset showed weaker effects likely due to limited statistical power.

Our findings reinforce existing research on the importance of sleep for academic performance and add value by demonstrating that the sleep–GPA association holds across populations and measurement methods, though effect sizes are modest and relationships are correlational rather than causal. These results should be interpreted as supporting general public health messaging around healthy sleep habits in college students rather than as a basis for individual prediction or intervention.

## Background and Prior Work

<p>Smartphones have become an almost inseparable part of college student's lives, and consequently, the habit of using them late into the night has become increasingly common. While there is lots of documentation on how poor sleep affects academic performance, less attention has been given to the broader relationship between smartphone usage patterns, sleep, and academic outcomes. College students need to balance between entertainment, messaging, and media scrolling with the sleep time they need to reinforce what they learned and to perform well academically. Excessive smartphone usage may contribute to delayed or inconsistent sleep schedules, which have been associated with negative health and cognitive outcomes in prior research.<a name="cite_ref-7"></a><a href="#cite_note-7"><sup>7</sup></a> Understanding how screen time behaviors relate to GPA and academic performance may therefore provide insight into how modern screen time habits impact student success.</p>

<p>Instead of directly measuring sleep, this project examines whether smartphone usage duration, sleep duration, and sleep consistency together correlate with quarterly GPA among college students. To do this we draw on multiple complementary public datasets: the StudentLife dataset provides objective sensor-based phone activity and sleep data paired with GPA for Dartmouth students; the CMU Sleep dataset provides Fitbit-measured sleep duration and consistency paired with GPA for a larger U.S. sample; the Global University Students Performance Habits dataset provides self-reported screen use and sleep alongside transcript GPA for a nationally representative U.S. sample; and the India Screen Time dataset provides daily screen time broken down by usage type alongside GPA across two semesters. Together these datasets allow us to triangulate the relationship between screen time, sleep, and academic performance across different populations and measurement methods. <a name="cite_ref-3"></a><a href="#cite_note-3"><sup>3</sup></a></p>

It is well established that the general usage of smartphones can cause physical and mental discomfort. It’s well known that excessive screentime is associated with concentration difficulties.<a name="cite_ref-5"></a><a href="#cite_note-5"><sup>5</sup></a> Electronic devices such as smartphones emit blue light, which is high in energy and short in wavelength. Many studies highlighted a negative relationship between bedtime phone usage and academic performance, with many students with less hours of sleep due to phone usage reporting a higher tiredness associated with low productivity.<a name="cite_ref-2"></a><a href="#cite_note-2"><sup>2</sup></a><a name="cite_ref-6"></a><a href="#cite_note-6"><sup>6</sup></a> It is also known that sleep deprivation affects adolescent development, causing impaired cognitive functioning. This leads to poorer decision making and decreased alertness, attention span, creativity, and working memory. These effects are caused by “reduced metabolic activity within the prefrontal cortex and other neural regions”.<a name="cite_ref-8"></a><a href="#cite_note-8"><sup>8</sup></a> These are extremely important for executive function, and as a result, sleep deficiency can lead to lower academic performance among students. Many of these studies used self reported data, which is limited by bias and validity. ### Dataset #1: StudentLife

This dataset comes from the a published academic study called StudentLife that tracked the daily behavior, mental health, and academic performance of college students using smartphone sensor data. The study collected continuous passive data (e.g., phone usage, environmental context) along with survey and academic records for a cohort of undergraduate students at Dartmouth College. Data was collected over an academic term using students’ personal smartphones, allowing for fine-grained observation of behavioral patterns. Academic performance, such as GPA, was obtained from institutional records, making it more reliable than self-reported measures.

The most important metrics in this dataset are as follows. The dark sensor records periods when the phone is in a dark environment, which can serve as a proxy for inactivity or potential sleep. The phonelock sensor captures when the phone is locked or unlocked, providing a basis for estimating phone usage behavior. The EMA (experience sampling) data contains self-reported responses throughout the day, including stress levels, mood, and sleep-related questions, typically recorded on ordinal scales. The education data includes academic outcomes such as GPA on a 0–4 scale, where higher values indicate stronger academic performance. Additional data sources, such as app_usage, conversation, and bluetooth, provide context about social interaction and daily behavior, although these are not directly used in all analyses.

A major concern with this dataset is that many key variables are not directly measured but must be inferred from sensor data. For example, sleep is not explicitly recorded and must be approximated from periods of phone inactivity and darkness, which can introduce measurement error. Similarly, phone usage is inferred from phone lock and unlock events, which may not capture all forms of interaction. The dataset is also subject to missing data, and this missingness may not be random. For instance, students who disable sensors or use their phones less frequently may systematically differ from those with more complete data. Additionally, the sample consists of students from a single university who voluntarily participated in a smartphone sensing study, which introduces self-selection bias and limits generalizability. Finally, while the dataset provides rich behavioral detail, it does not directly measure all aspects of interest (such as exact sleep or full digital engagement), requiring careful interpretation of derived variables.

```python
#StudentLife helper functions
def read_csv_if_exists(path):
    """Read a CSV if it exists, otherwise return None."""
    path = Path(path)
    if path.exists():
        return pd.read_csv(path)
    return None


def to_datetime_seconds(series):
    """Convert Unix seconds to pandas datetime."""
    return pd.to_datetime(pd.to_numeric(series, errors="coerce"), unit="s", errors="coerce")


def get_user_ids():
    """Find users from available phonelock files."""
    files = list((Path(f"{RAW_DIR}/dataset/sensing/phonelock")).glob("phonelock_u*.csv"))
    return sorted(file.stem.replace("phonelock_", "") for file in files)
```

```python
#Creates a interim, daily-user level, aggregated dataset from the multiple data files
all_daily_rows = []
user_ids = get_user_ids()

for user_id in user_ids:
    #Phone usage from phonelock
    lock_path = Path(f"{RAW_DIR}/dataset/sensing/phonelock/phonelock_{user_id}.csv")
    lock = read_csv_if_exists(lock_path)
    if lock is None or lock.empty:
        continue
    lock["start_dt"] = to_datetime_seconds(lock["start"])
    lock["end_dt"] = to_datetime_seconds(lock["end"])
    lock = lock.dropna(subset=["start_dt", "end_dt"])
    lock = lock[lock["end_dt"] > lock["start_dt"]]
    if lock.empty:
        continue
    lock["date"] = lock["start_dt"].dt.date
    lock["locked_hours"] = (lock["end_dt"] - lock["start_dt"]).dt.total_seconds() / 3600
    daily = lock.groupby("date", as_index=False)["locked_hours"].sum()

    #Assuming if the phone is locked, it is not actively being used
    daily["phone_usage_hours"] = (24 - daily["locked_hours"]).clip(lower=0, upper=24)
    daily = daily.drop(columns="locked_hours")
    daily["user_id"] = user_id

    # Sleep inferred from dark sensor
    dark_path = Path(f"{RAW_DIR}/dataset/sensing/dark/dark_{user_id}.csv")
    dark = read_csv_if_exists(dark_path)

    if dark is not None and not dark.empty:
        dark["start_dt"] = to_datetime_seconds(dark["start"])
        dark["end_dt"] = to_datetime_seconds(dark["end"])
        dark = dark.dropna(subset=["start_dt", "end_dt"])
        dark = dark[dark["end_dt"] > dark["start_dt"]]
        dark["sleep_hours"] = (dark["end_dt"] - dark["start_dt"]).dt.total_seconds()/3600

        #Estimated bounds of darkness for realistic sleep
        MIN_SLEEP_HOURS =  1
        MAX_SLEEP_HOURS = 12
        sleep_like = dark[
            (dark["sleep_hours"] >= MIN_SLEEP_HOURS) &
            (dark["sleep_hours"] <= MAX_SLEEP_HOURS)
        ].copy()

        # Assigned sleep to the day the sleep episode starts
        sleep_like["date"] = sleep_like["start_dt"].dt.date

        # Summed all valid sleep_like intervals that start on the same date
        sleep_daily = (
            sleep_like
            .groupby("date", as_index=False)["sleep_hours"]
            .sum()
        )
        sleep_daily["sleep_hours"] = sleep_daily["sleep_hours"].clip(lower=0, upper=16)
        daily = daily.merge(sleep_daily, on="date", how="left")

    else:
        daily["sleep_hours"] = np.nan

    # Stress from EMA responses
    stress_path = Path(f"{RAW_DIR}/dataset/EMA/response/Stress/Stress_{user_id}.json")
    stress_rows = []
    if stress_path.exists():
        with open(stress_path) as f:
            records = json.load(f)
        for record in records:
            if "resp_time" not in record:
                continue
            stress_value = record.get("level")
            if stress_value is None:
                stress_value = record.get("null")
            stress_rows.append({
                "date": pd.to_datetime(record["resp_time"], unit="s").date(),
                "stress_level": pd.to_numeric(stress_value, errors="coerce"),
            })
    if stress_rows:
        stress = pd.DataFrame(stress_rows).dropna(subset=["stress_level"])
        stress_daily = stress.groupby("date", as_index=False)["stress_level"].mean()
        daily = daily.merge(stress_daily, on="date", how="left")
    else:
        daily["stress_level"] = np.nan

    all_daily_rows.append(daily)

daily_behavior = pd.concat(all_daily_rows, ignore_index=True)
daily_behavior = daily_behavior[["user_id", "date", "sleep_hours", "phone_usage_hours", "stress_level"]]

print("Shape:", daily_behavior.shape)
print("Unique users:",daily_behavior["user_id"].nunique())
print("Unique days:",daily_behavior["date"].nunique())
print(daily_behavior.head())
```

```python
#Duplicates and missingness checking
print("Duplicate user-date rows:", daily_behavior.duplicated(["user_id", "date"]).sum())

#Overall missingness
print("Overall missingness:")
print(daily_behavior.isna().mean())

#Missingness by user
print("Missingness by user:")
missing_by_user = (daily_behavior.groupby("user_id")[["sleep_hours","stress_level"]].apply(lambda x: x.isna().mean()))
display(missing_by_user.describe())

#Missingness by date
print("Missingness by date:")
missing_by_date = (daily_behavior.groupby("date")[["sleep_hours","stress_level"]].apply(lambda x: x.isna().mean()))
display(missing_by_date.describe())

#Systematic check
print("Means with missing sleep included or not:")
display(daily_behavior.groupby(daily_behavior["sleep_hours"].isna())[["phone_usage_hours", "stress_level"]].mean())
```

Average missingness of sleep hours is relatively low while missingness of stress levels is very high. The variability of missingness of sleep by user is relatively high, but the variability by day is relatively low. On the other hand, the variability, in stress is high both by user and day showing that it is likely not random, and combined with the high degree of missingness likely will lead to bias. The column is chosen to be dropped for more clear analysis.

```python
#Dropping stress_level
daily_behavior = daily_behavior.drop(columns = ["stress_level"])
daily_behavior["flag_sleep_missing"] = daily_behavior["sleep_hours"].isna().astype(int)
daily_behavior["flag_phone_missing"] = daily_behavior["phone_usage_hours"].isna().astype(int)
daily_behavior["flag_low_sleep_coverage"] = (daily_behavior.groupby("user_id")["sleep_hours"].transform(lambda x: x.notna().mean()) < 0.5).astype(int)

#Save interim dataset
daily_behavior.to_csv(f"{INTERIM_DIR}/daily_behavior.csv", index=False)
print("Saved interim dataset:", f"{INTERIM_DIR}/daily_behavior.csv")
```

```python
#Aggregate user-day level data to user level data
user_behavior = (daily_behavior.groupby("user_id", as_index=False).agg(avg_sleep_hours=("sleep_hours", "mean"),\
    avg_phone_usage_hours=("phone_usage_hours", "mean"), days_observed=("date", "nunique"),percent_flag_sleep_missing=("flag_sleep_missing", "mean"),\
    percent_flag_phone_missing=("flag_phone_missing", "mean"), percent_flag_low_sleep=("flag_low_sleep_coverage", "mean")))

#Combine with academic performance data
grades = read_csv_if_exists(Path(f"{RAW_DIR}/dataset/education/grades.csv"))
grades = grades.rename(columns={"uid": "user_id"})
grades.columns = grades.columns.str.strip().str.lower().str.replace(" ", "_")
user_behavior = user_behavior.merge(grades, on="user_id", how="left")

# Deadlines included as potential confounder for performance, and proxy for amount of coursework
deadlines = read_csv_if_exists(Path(f"{RAW_DIR}/dataset/education/deadlines.csv"))
if deadlines is not None:
    deadlines = deadlines.rename(columns={"uid": "user_id"})
    deadline_long = deadlines.melt(id_vars="user_id", var_name="date", value_name="deadline_count")
    deadline_long["deadline_count"] = pd.to_numeric(deadline_long["deadline_count"], errors="coerce").fillna(0)
    workload = deadline_long.groupby("user_id", as_index=False)["deadline_count"].mean()
    workload = workload.rename(columns={"deadline_count": "avg_deadline_count"})
    user_behavior = user_behavior.merge(workload, on="user_id", how="left")
```

### Dataset #4: CMU Sleep Study: Sleep Duration, Consistency, and GPA

This dataset comes from a published academic study tracking the sleep behavior and academic performance of 634 first-year college students across three U.S. universities: Carnegie Mellon University, University of Washington, and University of Notre Dame. Sleep was recorded using Fitbit wearable devices over approximately one month during the spring academic term, and GPA records were obtained directly from university registrars, making the academic performance measure more reliable than self-reported alternatives.

The most important metrics in this dataset are as follows. `TotalSleepTime` measures average nightly sleep duration in minutes — healthy sleep for young adults falls between 420 and 540 minutes (7–9 hours), and values below 360 minutes (6 hours) have been associated with significant cognitive impairment and lower academic performance. `bedtime_mssd` is the mean successive squared difference of bedtime in minutes squared, measuring night-to-night variability in sleep timing; a higher value indicates more irregular sleep schedules, which is directly relevant to our consistency hypothesis. `midpoint_sleep` is the average midpoint of the sleep episode measured in minutes after 11PM — for example, a value of 240 corresponds to a sleep midpoint of 3AM, indicating a late sleep schedule that overlaps with our 11PM–6AM window of interest. `daytime_sleep` captures average minutes of sleep outside the main nightly episode; elevated values may suggest chronic sleep deprivation being compensated through napping. `term_gpa` and `cum_gpa` both measure academic performance on a 4.0 scale, with `term_gpa` reflecting the specific term during which sleep was tracked and `cum_gpa` reflecting prior academic history.

A major concern with this dataset is that it contains no smartphone screen time data whatsoever, meaning it can only address the sleep → GPA portion of our research question and not the screen time → GPA portion directly. Additionally, the sample is restricted to first-year students, who may behave differently from upperclassmen in terms of sleep habits and academic adjustment. The study also only tracked sleep for roughly one month rather than the full academic quarter, which may not capture the full range of sleep behavior over a term. Finally, participants were drawn from only three universities, limiting generalizability to broader college populations.

## Discussion and Conclusion

This project investigated whether daily sleep duration and smartphone usage are associated with quarterly academic performance, measured by GPA, among university students. To address this question we drew on four complementary datasets: the StudentLife dataset (Dartmouth, n=48), the Global University Students Performance Habits dataset (multi-country, n=10,000), the Student Screen Time vs CGPA dataset (India, n=550), and the CMU Sleep dataset (CMU/UW/Notre Dame, n=634). Each dataset captured different aspects of the relationship between sleep, screen time, and academic outcomes, allowing us to triangulate findings across populations and measurement methods rather than relying on a single source.

Across the four datasets, the evidence pointed in the same general direction but with notable variation in strength. The CMU Sleep dataset showed a statistically significant positive association between sleep duration and GPA — students in the high-sleep group earned a GPA roughly 0.30 points higher than those in the low-sleep group (p ≈ 0.004), and this effect remained even after controlling for gender, race, first-generation student status, and units enrolled. The Global University Students Performance Habits dataset, with its much larger sample size, similarly identified associations between both sleep and screen time with GPA. The India-based Screen Time vs CGPA dataset corroborated the screen time → GPA portion of these findings independently. The StudentLife dataset showed weaker or non-significant relationships, though its small sample size and uneven tracking durations across participants limit the reliability of those findings.

These results extend the prior work discussed in our background section in two important ways. Prior studies established that bedtime smartphone use, blue light exposure, and irregular sleep are each individually associated with reduced sleep quality and academic performance. Our project does not contradict that body of literature, but it adds an empirical comparison across multiple datasets that vary substantially in measurement method (Fitbit-tracked vs self-reported), population (U.S. selective universities vs Indian undergraduates vs global samples), and sample size (48 to 10,000). The fact that we observed broadly consistent directionality across such different data sources strengthens the overall claim that sleep and screen time meaningfully relate to GPA, while also highlighting that the magnitude of these relationships is sensitive to how the variables are measured and which population is studied.

The results from our analyses in three of the four datasets show that sleep time is positively related to GPA and screen time is negatively associated with GPA. This is consistent with our hypothesis, and resonates with common sense. The results act as good advice for students who are seeking to level up their academic performance: cutting down sleep time to study more might not necessarily help with one’s academics, and excessive screen time is harmful. Even though dataset 1 found no apparent relation between sleep, screen time and GPA, in terms of developing study habits and students’ time management, what we learn from datasets 2, 3 and 4 are still informative.

There are some contradictory results in our analyses. Dataset 1 found no apparent relation between sleep, screen time, and GPA, with a low correlation and high p-value even in the simple linear regression models. Dataset 2 found an association between sleep, screen time, and GPA that was consistent with our hypothesis. Dataset 3 and 4 also found support conclusions that are consistent with our hypothesis. The contradiction might be due to the fact that dataset 1 is a small dataset of only 30 observations, and it has its own biases and limitations. This small dataset might have some confounding factors specific to it that generated extra noise. However, it is worth noting that dataset 1 does not necessarily disprove the results obtained from datasets 2, 3 and 4. The conclusions from those three datasets still hold individually. It is just that overall we cannot draw the conclusion with full certainty that our hypothesis is true.

The biggest limitation of this analysis is that it cannot be generalized to our target population of college students in the US for the reason that all our datasets had some flaws. Dataset 1 contained information on our population and the exact things we were measuring, however it was very small, making us uncertain of the results and unable to generalize our results. Dataset 2 included data from countries other than the US, so we cannot generalize it for that reason. Another issue is that we don’t know how some of our data was measured and if it was collected in an unbiased way. All of our datasets also do not cover the same population so we cannot directly compare them. We are also using GPA as a proxy measurement for academic performance, and so it may not fully capture how a student is truly performing. Students may be able to improve their GPA without actually improving their learning or understanding of the material and vice versa.

More data from the population we are trying to measure would definitely make us much more confident in any result we get. In addition, longitudinal data would help more, because we are basing off our analysis on a single measurement (GPA, Sleep hours, screentime), which can fluctuate for many different reasons.

# Research Question

Does the average daily sleep and the average length of daily smartphone usage of university students have an association with overall quarterly academic performance, measured by average quarter GPA?

## Background and Prior Work

<p>Smartphones have become an almost inseparable part of college student's lives, and consequently, the habit of using them late into the night has become increasingly common. While there is lots of documentation on how poor sleep affects academic performance, less attention has been given to the broader relationship between smartphone usage patterns, sleep, and academic outcomes. College students need to balance between entertainment, messaging, and media scrolling with the sleep time they need to reinforce what they learned and to perform well academically. Excessive smartphone usage may contribute to delayed or inconsistent sleep schedules, which have been associated with negative health and cognitive outcomes in prior research.<a name="cite_ref-7"></a><a href="#cite_note-7"><sup>7</sup></a> Understanding how screen time behaviors relate to GPA and academic performance may therefore provide insight into how modern screen time habits impact student success.</p>

<p>Instead of directly measuring sleep, this project examines whether smartphone usage duration, sleep duration, and sleep consistency together correlate with quarterly GPA among college students. To do this we draw on multiple complementary public datasets: the StudentLife dataset provides objective sensor-based phone activity and sleep data paired with GPA for Dartmouth students; the CMU Sleep dataset provides Fitbit-measured sleep duration and consistency paired with GPA for a larger U.S. sample; the Global University Students Performance Habits dataset provides self-reported screen use and sleep alongside transcript GPA for a nationally representative U.S. sample; and the India Screen Time dataset provides daily screen time broken down by usage type alongside GPA across two semesters. Together these datasets allow us to triangulate the relationship between screen time, sleep, and academic performance across different populations and measurement methods. <a name="cite_ref-3"></a><a href="#cite_note-3"><sup>3</sup></a></p>

It is well established that the general usage of smartphones can cause physical and mental discomfort. It’s well known that excessive screentime is associated with concentration difficulties.<a name="cite_ref-5"></a><a href="#cite_note-5"><sup>5</sup></a> Electronic devices such as smartphones emit blue light, which is high in energy and short in wavelength. Long exposure to this affects sleep by inhibiting melatonin production, creating circadian disruptions, leading to a deterioration of both sleep quality and duration.<a name="cite_ref-1"></a><a href="#cite_note-1"><sup>1</sup></a> With the prevalence of electronic devices for adolescents, there have been many studies on whether smart phone usage has an impact on academic performance. Evidence has suggested that increased screen time may reduce the "functional connectivity between cognitive areas"<a name="cite_ref-8"></a><a href="#cite_note-8"><sup>8</sup></a>, however it's direct effect on academic performance is disputed. Many studies highlighted a negative relationship between bedtime phone usage and academic performance, with many students with less hours of sleep due to phone usage reporting a higher tiredness associated with low productivity.<a name="cite_ref-2"></a><a href="#cite_note-2"><sup>2</sup></a><a name="cite_ref-6"></a><a href="#cite_note-6"><sup>6</sup></a> It is also known that sleep deprivation affects adolescent development, causing impaired cognitive functioning. This leads to poorer decision making and decreased alertness, attention span, creativity, and working memory. These effects are caused by “reduced metabolic activity within the prefrontal cortex and other neural regions”.<a name="cite_ref-8"></a><a href="#cite_note-8"><sup>8</sup></a> These are extremely important for executive function, and as a result, sleep deficiency can lead to lower academic performance among students. Many of these studies used self reported data, which is limited by bias and validity. They also ask students to report how they feel rather than use an arbitrary measurement, which can also increase the likelihood of bias.<a name="cite_ref-3"></a><a href="#cite_note-3"><sup>3</sup></a> We want to use screentime and the average amount of sleep to determine if there is an association with overall quarterly academic performance, measured by average quarter GPA.

### Dataset #1: StudentLife

This dataset comes from the a published academic study called StudentLife that tracked the daily behavior, mental health, and academic performance of college students using smartphone sensor data. The study collected continuous passive data (e.g., phone usage, environmental context) along with survey and academic records for a cohort of undergraduate students at Dartmouth College. Data was collected over an academic term using students’ personal smartphones, allowing for fine-grained observation of behavioral patterns. Academic performance, such as GPA, was obtained from institutional records, making it more reliable than self-reported measures.

The most important metrics in this dataset are as follows. The dark sensor records periods when the phone is in a dark environment, which can serve as a proxy for inactivity or potential sleep. The phonelock sensor captures when the phone is locked or unlocked, providing a basis for estimating phone usage behavior. The EMA (experience sampling) data contains self-reported responses throughout the day, including stress levels, mood, and sleep-related questions, typically recorded on ordinal scales. The education data includes academic outcomes such as GPA on a 0–4 scale, where higher values indicate stronger academic performance. Additional data sources, such as app_usage, conversation, and bluetooth, provide context about social interaction and daily behavior, although these are not directly used in all analyses.

A major concern with this dataset is that many key variables are not directly measured but must be inferred from sensor data. For example, sleep is not explicitly recorded and must be approximated from periods of phone inactivity and darkness, which can introduce measurement error. Similarly, phone usage is inferred from phone lock and unlock events, which may not capture all forms of interaction. The dataset is also subject to missing data, and this missingness may not be random. For instance, students who disable sensors or use their phones less frequently may systematically differ from those with more complete data. Additionally, the sample consists of students from a single university who voluntarily participated in a smartphone sensing study, which introduces self-selection bias and limits generalizability. Finally, while the dataset provides rich behavioral detail, it does not directly measure all aspects of interest (such as exact sleep or full digital engagement), requiring careful interpretation of derived variables.

```python
#StudentLife helper functions
def read_csv_if_exists(path):
    """Read a CSV if it exists, otherwise return None."""
    path = Path(path)
    if path.exists():
        return pd.read_csv(path)
    return None


def to_datetime_seconds(series):
    """Convert Unix seconds to pandas datetime."""
    return pd.to_datetime(pd.to_numeric(series, errors="coerce"), unit="s", errors="coerce")


def get_user_ids():
    """Find users from available phonelock files."""
    files = list((Path(f"{RAW_DIR}/dataset/sensing/phonelock")).glob("phonelock_u*.csv"))
    return sorted(file.stem.replace("phonelock_", "") for file in files)
```

```python
#Creates a interim, daily-user level, aggregated dataset from the multiple data files
all_daily_rows = []
user_ids = get_user_ids()

for user_id in user_ids:
    #Phone usage from phonelock
    lock_path = Path(f"{RAW_DIR}/dataset/sensing/phonelock/phonelock_{user_id}.csv")
    lock = read_csv_if_exists(lock_path)
    if lock is None or lock.empty:
        continue
    lock["start_dt"] = to_datetime_seconds(lock["start"])
    lock["end_dt"] = to_datetime_seconds(lock["end"])
    lock = lock.dropna(subset=["start_dt", "end_dt"])
    lock = lock[lock["end_dt"] > lock["start_dt"]]
    if lock.empty:
        continue
    lock["date"] = lock["start_dt"].dt.date
    lock["locked_hours"] = (lock["end_dt"] - lock["start_dt"]).dt.total_seconds() / 3600
    daily = lock.groupby("date", as_index=False)["locked_hours"].sum()

    #Assuming if the phone is locked, it is not actively being used
    daily["phone_usage_hours"] = (24 - daily["locked_hours"]).clip(lower=0, upper=24)
    daily = daily.drop(columns="locked_hours")
    daily["user_id"] = user_id

    # Sleep inferred from dark sensor
    dark_path = Path(f"{RAW_DIR}/dataset/sensing/dark/dark_{user_id}.csv")
    dark = read_csv_if_exists(dark_path)

    if dark is not None and not dark.empty:
        dark["start_dt"] = to_datetime_seconds(dark["start"])
        dark["end_dt"] = to_datetime_seconds(dark["end"])
        dark = dark.dropna(subset=["start_dt", "end_dt"])
        dark = dark[dark["end_dt"] > dark["start_dt"]]
        dark["sleep_hours"] = (dark["end_dt"] - dark["start_dt"]).dt.total_seconds()/3600

        #Estimated bounds of darkness for realistic sleep
        MIN_SLEEP_HOURS =  1
        MAX_SLEEP_HOURS = 12
        sleep_like = dark[
            (dark["sleep_hours"] >= MIN_SLEEP_HOURS) &
            (dark["sleep_hours"] <= MAX_SLEEP_HOURS)
        ].copy()

        # Assigned sleep to the day the sleep episode starts
        sleep_like["date"] = sleep_like["start_dt"].dt.date

        # Summed all valid sleep_like intervals that start on the same date
        sleep_daily = (
            sleep_like
            .groupby("date", as_index=False)["sleep_hours"]
            .sum()
        )
        sleep_daily["sleep_hours"] = sleep_daily["sleep_hours"].clip(lower=0, upper=16)
        daily = daily.merge(sleep_daily, on="date", how="left")

    else:
        daily["sleep_hours"] = np.nan

    # Stress from EMA responses
    stress_path = Path(f"{RAW_DIR}/dataset/EMA/response/Stress/Stress_{user_id}.json")
    stress_rows = []
    if stress_path.exists():
        with open(stress_path) as f:
            records = json.load(f)
        for record in records:
            if "resp_time" not in record:
                continue
            stress_value = record.get("level")
            if stress_value is None:
                stress_value = record.get("null")
            stress_rows.append({
                "date": pd.to_datetime(record["resp_time"], unit="s").date(),
                "stress_level": pd.to_numeric(stress_value, errors="coerce"),
            })
    if stress_rows:
        stress = pd.DataFrame(stress_rows).dropna(subset=["stress_level"])
        stress_daily = stress.groupby("date", as_index=False)["stress_level"].mean()
        daily = daily.merge(stress_daily, on="date", how="left")
    else:
        daily["stress_level"] = np.nan

    all_daily_rows.append(daily)

daily_behavior = pd.concat(all_daily_rows, ignore_index=True)
daily_behavior = daily_behavior[["user_id", "date", "sleep_hours", "phone_usage_hours", "stress_level"]]

print("Shape:", daily_behavior.shape)
print("Unique users:",daily_behavior["user_id"].nunique())
print("Unique days:",daily_behavior["date"].nunique())
print(daily_behavior.head())
```

```python
#Duplicates and missingness checking
print("Duplicate user-date rows:", daily_behavior.duplicated(["user_id", "date"]).sum())

#Overall missingness
print("Overall missingness:")
print(daily_behavior.isna().mean())

#Missingness by user
print("Missingness by user:")
missing_by_user = (daily_behavior.groupby("user_id")[["sleep_hours","stress_level"]].apply(lambda x: x.isna().mean()))
display(missing_by_user.describe())

#Missingness by date
print("Missingness by date:")
missing_by_date = (daily_behavior.groupby("date")[["sleep_hours","stress_level"]].apply(lambda x: x.isna().mean()))
display(missing_by_date.describe())

#Systematic check
print("Means with missing sleep included or not:")
display(daily_behavior.groupby(daily_behavior["sleep_hours"].isna())[["phone_usage_hours", "stress_level"]].mean())
```

Average missingness of sleep hours is relatively low while missingness of stress levels is very high. The variability of missingness of sleep by user is relatively high, but the variability by day is relatively low. On the other hand, the variability, in stress is high both by user and day showing that it is likely not random, and combined with the high degree of missingness likely will lead to bias. ### Dataset #4: CMU Sleep Study: Sleep Duration, Consistency, and GPA

This dataset comes from a published academic study tracking the sleep behavior and academic performance of 634 first-year college students across three U.S. universities: Carnegie Mellon University, University of Washington, and University of Notre Dame. Sleep was recorded using Fitbit wearable devices over approximately one month during the spring academic term, and GPA records were obtained directly from university registrars, making the academic performance measure more reliable than self-reported alternatives.

The most important metrics in this dataset are as follows. `TotalSleepTime` measures average nightly sleep duration in minutes — healthy sleep for young adults falls between 420 and 540 minutes (7–9 hours), and values below 360 minutes (6 hours) have been associated with significant cognitive impairment and lower academic performance. `bedtime_mssd` is the mean successive squared difference of bedtime in minutes squared, measuring night-to-night variability in sleep timing; a higher value indicates more irregular sleep schedules, which is directly relevant to our consistency hypothesis. `midpoint_sleep` is the average midpoint of the sleep episode measured in minutes after 11PM — for example, a value of 240 corresponds to a sleep midpoint of 3AM, indicating a late sleep schedule that overlaps with our 11PM–6AM window of interest. `daytime_sleep` captures average minutes of sleep outside the main nightly episode; elevated values may suggest chronic sleep deprivation being compensated through napping. `term_gpa` and `cum_gpa` both measure academic performance on a 4.0 scale, with `term_gpa` reflecting the specific term during which sleep was tracked and `cum_gpa` reflecting prior academic history.

A major concern with this dataset is that it contains no smartphone screen time data whatsoever, meaning it can only address the sleep → GPA portion of our research question and not the screen time → GPA portion directly. Additionally, the sample is restricted to first-year students, who may behave differently from upperclassmen in terms of sleep habits and academic adjustment. The study also only tracked sleep for roughly one month rather than the full academic quarter, which may not capture the full range of sleep behavior over a term. Finally, participants were drawn from only three universities, limiting generalizability to broader college populations.

##### StudentLife Dataset: Confounding Variable Analysis

Our earlier analysis has shown no significant GPA and sleep or GPA and screen usage relationships, but this could be partly due to confounders masking relationships. This section will check if accounting for confounders in the dataset impacts evidence of these relationships. First we check the significance of whether these confounders are correlated with our key variables.

```python
# ── Deadline count as a confounder ──────────────────────────────────────
# avg_deadline_count proxies workload, which could affect both sleep and GPA
confounders = ['avg_deadline_count', 'days_observed']
vars_of_interest = ['avg_sleep_hours', 'avg_phone_usage_hours', 'gpa_all']

print("\n=== Correlations: Confounders vs Key Variables ===")
print("Spearman r (rank-based — preferred for skewed vars)")
print("Spearman is the primary metric here since avg_deadline_count is highly skewed.")
for conf in confounders:
    print(f"\n--- {conf} ---")
    for var in vars_of_interest:
        mask = user_behavior[[conf, var]].dropna()
        r_s, p_s = stats.spearmanr(mask[conf], mask[var])
        print(f"  vs {var}: Spearman r={r_s:.3f}, p={p_s:.4f}")

# ── Scatter plots: Confounders vs key variables ─────────────────────
# Visual check of any linear or non-linear relationships with confounders.
fig, axes = plt.subplots(len(confounders), len(vars_of_interest),
                         figsize=(14, 8))
for i, conf in enumerate(confounders):
    for j, var in enumerate(vars_of_interest):
        ax = axes[i][j]
        valid = user_behavior[[conf, var]].dropna()
        ax.scatter(valid[conf], valid[var], alpha=0.6, color='steelblue', s=60)
        ax.set_xlabel(conf)
        ax.set_ylabel(var)
        r_s, p_s = stats.spearmanr(valid[conf], valid[var])
        ax.set_title(f'{conf} vs {var}\nSpearman r={r_s:.3f}', fontsize=9)
plt.suptitle('Confounding Variable Analysis — StudentLife Dataset', fontsize=14)
plt.tight_layout()
plt.show()
```

p-values from confounders seem plausible under no significant association. We now perform a multiple regression to distinguish the effects of sleep and screen time usage on GPA, using condition number to keep track of the multicolinearity issue. We transform avg_deadline_count using a log transform as well to manage the much larger skew present in that variable.

```python
# ── Multiple regression: controlling for confounders ────────────────────
# Does sleep still predict GPA after controlling for deadline count and days observed?
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler

valid_mlr = user_behavior[['avg_sleep_hours', 'avg_phone_usage_hours',
                          'avg_deadline_count', 'days_observed', 'gpa_all']].dropna().copy()

valid_mlr['log_deadline'] = np.log(valid_mlr['avg_deadline_count'])

X = valid_mlr[['avg_sleep_hours', 'avg_phone_usage_hours',
               'log_deadline', 'days_observed']]
y = valid_mlr['gpa_all']

mlr = LinearRegression().fit(X, y)
y_pred = mlr.predict(X)

n, p = X.shape
r2 = r2_score(y, y_pred)
adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)

# Condition number on standardized predictors (scale-invariant)
X_scaled = StandardScaler().fit_transform(X)
cond = np.linalg.cond(X_scaled)

print("\n=== Multiple Linear Regression: Predicting GPA ===")
print(f"R² = {r2:.4f}")
print(f"Adjusted R² = {adj_r2:.4f}")
print(f"Condition number (stdized) = {cond:.1f}  (threshold: > 30 is problematic)")
for name, coef in zip(X.columns, mlr.coef_):
    print(f"  {name}: coef = {coef:.4f}")
print(f"  intercept: {mlr.intercept_:.4f}")
```

The condition number is quite low and stable, but the R^2 and adjusted R^2 are quite low, indicating that the predictors do not explain very little variance in the GPA. We were not able to find a clearly apparent relationship in the StudentLife dataset between sleep, screen time, and GPA.

##### **Permutation Tests**

To test our research hypothesis with this dataset, we decided to conduct two **binned permutation tests**: `sleep_hours` and `GPA`, `screen_time_hours` and `GPA`:

**Null hypothesis**: `GPA` of students in different `sleep_hours` bins are the **same** / `GPA` of students in different `screen_time_hours` bins are the **same**.

**Alternative hypothesis**: `GPA` of students in different `sleep_hours` bins are **different** / `GPA` of students in different `screen_time_hours` bins are **different**.

```python
# binning sleep_hours and screen_time_hours
data2['sleep_groups'] = pd.cut(data2['sleep_hours'], bins=[0, 6, 8, data2['sleep_hours'].max()], labels=['less than 6 hrs', 'between 6 and 8 hrs', 'more than 8 hrs'])
data2['screen_time_groups'] = pd.cut(data2['screen_time_hours'], bins=[0, 3, 6, data2['screen_time_hours'].max()], labels=['less than 3 hrs', 'between 3 and 6 hrs', 'more than 6 hrs'])
```

```python
# sleep_hours_group and GPA test
# Using the variance of grouped means as the test statistic
observed = data2.groupby('sleep_groups', observed=True)['GPA'].mean().var()
simulated = []
for _ in range(10000):
    with_shuffle = data2.assign(shuffled_GPA = np.random.permutation(data2['GPA']))
    stat = with_shuffle.groupby('sleep_groups', observed=True)['shuffled_GPA'].mean().var()
    simulated.append(stat)

simulated = np.array(simulated)
print(f'observed: {observed}, simulated: {simulated}')
print(f'p-value: {(simulated >= observed).mean()}')
```

```python
# screen_hours_group and GPA test
# Using the variance of grouped means as the test statistic
observed = data2.groupby('screen_time_groups', observed=True)['GPA'].mean().var()
simulated = []
for _ in range(10000):
    with_shuffle = data2.assign(shuffled_GPA = np.random.permutation(data2['GPA']))
    stat = with_shuffle.groupby('screen_time_groups', observed=True)['shuffled_GPA'].mean().var()
    simulated.append(stat)

simulated = np.array(simulated)
print(f'observed: {observed}, simulated: {simulated}')
print(f'p-value: {(simulated >= observed).mean()}')
```

From the p-value of the permutation tests, we can **reject the null**. The permutation tests **show evidence** that sleep hours and GPA are related, and so are screen hours and GPA. However, we need to a little bit more investigation using multiple linear regression to control for the other variables that might be confounding.

#### CMU Sleep Dataset: Multivariable Relational Analysis

We now examine the relationships between our key predictors (sleep duration and sleep consistency) and our outcome variable (term GPA). We use binned permutation tests to assess statistical significance and simple linear regression to quantify the direction and strength of each relationship.

```python
# ── Sleep duration vs GPA — binned permutation test ─────────────────
sleep_col = 'avg_nightly_sleep_minutes'
cmu_clean['sleep_group'] = pd.cut(cmu_clean[sleep_col], 
                                   bins=3, 
                                   labels=['Low Sleep', 'Medium Sleep', 'High Sleep'])

low_gpa = cmu_clean[cmu_clean['sleep_group'] == 'Low Sleep']['term_gpa'].dropna()
high_gpa = cmu_clean[cmu_clean['sleep_group'] == 'High Sleep']['term_gpa'].dropna()
observed_diff_sleep = high_gpa.mean() - low_gpa.mean()
print(f"Observed GPA difference (High - Low sleep): {observed_diff_sleep:.4f}")

n_permutations = 1000
combined = np.concatenate([low_gpa, high_gpa])
perm_diffs = []

for _ in range(n_permutations):
    shuffled = combined.copy()
    np.random.shuffle(shuffled)
    perm_diffs.append(shuffled[len(low_gpa):].mean() - shuffled[:len(low_gpa)].mean())

p_value_sleep = np.mean(np.abs(perm_diffs) >= np.abs(observed_diff_sleep))
print(f"Permutation test p-value (sleep vs GPA): {p_value_sleep:.4f}")

plt.figure(figsize=(8, 5))
plt.hist(perm_diffs, bins=40, color='steelblue', edgecolor='white', alpha=0.8)
plt.axvline(observed_diff_sleep, color='red', linestyle='--', 
            label=f'Observed diff: {observed_diff_sleep:.3f}')
plt.xlabel('Permuted GPA Difference')
plt.ylabel('Count')
plt.title('Permutation Test: Sleep Duration vs GPA — CMU Sleep')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ── Boxplot of GPA by sleep group ────────────────────────────────────
plt.figure(figsize=(8, 5))
cmu_clean.boxplot(column='term_gpa', by='sleep_group')
plt.suptitle('')
plt.title('Term GPA by Sleep Group — CMU Sleep Dataset')
plt.xlabel('Sleep Group')
plt.ylabel('Term GPA')
plt.tight_layout()
plt.show()
```

```python
# ── Sleep consistency vs GPA — binned permutation test ───────────────
consist_col = 'bedtime_night_to_night_variability'
cmu_clean['consistency_group'] = pd.cut(cmu_clean[consist_col],
                                         bins=3,
                                         labels=['Consistent', 'Moderate', 'Inconsistent'])

consistent_gpa = cmu_clean[cmu_clean['consistency_group'] == 'Consistent']['term_gpa'].dropna()
inconsistent_gpa = cmu_clean[cmu_clean['consistency_group'] == 'Inconsistent']['term_gpa'].dropna()
observed_diff_consist = consistent_gpa.mean() - inconsistent_gpa.mean()
print(f"Observed GPA difference (Consistent - Inconsistent): {observed_diff_consist:.4f}")

combined2 = np.concatenate([consistent_gpa, inconsistent_gpa])
perm_diffs2 = []

for _ in range(n_permutations):
    shuffled2 = combined2.copy()
    np.random.shuffle(shuffled2)
    perm_diffs2.append(shuffled2[:len(consistent_gpa)].mean() - shuffled2[len(consistent_gpa):].mean())

p_value_consist = np.mean(np.abs(perm_diffs2) >= np.abs(observed_diff_consist))
print(f"Permutation test p-value (consistency vs GPA): {p_value_consist:.4f}")

plt.figure(figsize=(8, 5))
plt.hist(perm_diffs2, bins=40, color='orange', edgecolor='white', alpha=0.8)
plt.axvline(observed_diff_consist, color='red', linestyle='--',
            label=f'Observed diff: {observed_diff_consist:.3f}')
plt.xlabel('Permuted GPA Difference')
plt.ylabel('Count')
plt.title('Permutation Test: Sleep Consistency vs GPA — CMU Sleep')
plt.legend()
plt.tight_layout()
plt.show()
```

#### CMU Sleep Dataset: Confounding Variable Analysis

Several demographic and academic variables in this dataset could confound the sleep-GPA relationship. We examine whether gender, race, first-generation student status, and units enrolled are correlated with our key predictors and outcome, and use multiple linear regression to test whether the sleep-GPA association holds after controlling for these factors.

```python
# ── Correlations between confounders and key variables ───────────────
confounders = ['gender', 'race', 'is_first_gen_student', 'units_enrolled']
confounders = [c for c in confounders if c in cmu_clean.columns]

# Convert confounders to numeric, coercing non-numeric values to NaN
for col in confounders:
    cmu_clean[col] = pd.to_numeric(cmu_clean[col], errors='coerce')

vars_of_interest = ['avg_nightly_sleep_minutes',
                    'bedtime_night_to_night_variability',
                    'term_gpa']

print("=== Correlations: Confounders vs Key Variables ===")
for conf in confounders:
    print(f"\n--- {conf} ---")
    for var in vars_of_interest:
        try:
            valid = cmu_clean[[conf, var]].dropna()
            r, p = stats.pearsonr(valid[conf], valid[var])
            print(f"  vs {var}: r={r:.3f}, p={p:.4f}")
        except Exception as e:
            print(f"  vs {var}: could not compute ({e})")
```

```python
# ── Multiple linear regression controlling for confounders ────────────
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

reg_vars = ['avg_nightly_sleep_minutes', 
            'bedtime_night_to_night_variability'] + confounders
reg_data = cmu_clean[reg_vars + ['term_gpa']].copy()

# Convert all columns to numeric, coercing any non-numeric values to NaN
for col in reg_data.columns:
    reg_data[col] = pd.to_numeric(reg_data[col], errors='coerce')

reg_data = reg_data.dropna()
print(f"Rows used in regression: {len(reg_data)}")

X = reg_data[reg_vars]
y = reg_data['term_gpa']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = LinearRegression()
model.fit(X_scaled, y)

print("\n=== Multiple Linear Regression: Predicting Term GPA ===")
print(f"R² score: {model.score(X_scaled, y):.4f}")
print("\nStandardized Coefficients:")
for name, coef in zip(reg_vars, model.coef_):
    print(f"  {name}: {coef:.4f}")
```

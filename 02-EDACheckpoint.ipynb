# Research Question

Does the average daily sleep and the average length of daily smartphone usage of university students have an association with overall quarterly academic performance, measured by average quarter GPA?

## Background and Prior Work

<p>Smartphones have become an almost inseparable part of college student's lives, and consequently, the habit of using them late into the night has become increasingly common. While there is lots of documentation on how poor sleep affects academic performance, less attention has been given to the broader relationship between smartphone usage patterns, sleep, and academic outcomes. College students need to balance between entertainment, messaging, and media scrolling with the sleep time they need to reinforce what they learned and to perform well academically. Excessive smartphone usage may contribute to delayed or inconsistent sleep schedules, which have been associated with negative health and cognitive outcomes in prior research.<a name="cite_ref-7"></a><a href="#cite_note-7"><sup>7</sup></a> Understanding how screen time behaviors relate to GPA and academic performance may therefore provide insight into how modern screen time habits impact student success.</p>

<p>Instead of directly measuring sleep, this project examines whether smartphone usage duration, sleep duration, and sleep consistency together correlate with quarterly GPA among college students. To do this we draw on multiple complementary public datasets: the StudentLife dataset provides objective sensor-based phone activity and sleep data paired with GPA for Dartmouth students; the CMU Sleep dataset provides Fitbit-measured sleep duration and consistency paired with GPA for a larger U.S. sample; the Global University Students Performance Habits dataset provides self-reported screen use and sleep alongside transcript GPA for a nationally representative U.S. sample; and the India Screen Time dataset provides daily screen time broken down by usage type alongside GPA across two semesters. Together these datasets allow us to triangulate the relationship between screen time, sleep, and academic performance across different populations and measurement methods. <a name="cite_ref-3"></a><a href="#cite_note-3"><sup>3</sup></a></p>

It is well established that the general usage of smartphones can cause physical and mental discomfort. It’s well known that excessive screentime is associated with concentration difficulties.<a name="cite_ref-5"></a><a href="#cite_note-5"><sup>5</sup></a> Electronic devices such as smartphones emit blue light, which is high in energy and short in wavelength. Long exposure to this affects sleep by inhibiting melatonin production, creating circadian disruptions, leading to a deterioration of both sleep quality and duration.<a name="cite_ref-1"></a><a href="#cite_note-1"><sup>1</sup></a> With the prevalence of electronic devices for adolescents, there have been many studies on whether smart phone usage has an impact on academic performance. Evidence has suggested that increased screen time may reduce the "functional connectivity between cognitive areas"<a name="cite_ref-8"></a><a href="#cite_note-8"><sup>8</sup></a>, however it's direct effect on academic performance is disputed. Many studies highlighted a negative relationship between bedtime phone usage and academic performance, with many students with less hours of sleep due to phone usage reporting a higher tiredness associated with low productivity.<a name="cite_ref-2"></a><a href="#cite_note-2"><sup>2</sup></a><a name="cite_ref-6"></a><a href="#cite_note-6"><sup>6</sup></a> It is also known that sleep deprivation affects adolescent development, causing impaired cognitive functioning. This leads to poorer decision making and decreased alertness, attention span, creativity, and working memory. These effects are caused by “reduced metabolic activity within the prefrontal cortex and other neural regions”.<a name="cite_ref-8"></a><a href="#cite_note-8"><sup>8</sup></a> These are extremely important for executive function, and as a result, sleep deficiency can lead to lower academic performance among students. Many of these studies used self reported data, which is limited by bias and validity. They also ask students to report how they feel rather than use an arbitrary measurement, which can also increase the likelihood of bias.<a name="cite_ref-3"></a><a href="#cite_note-3"><sup>3</sup></a> We want to use screentime and the average amount of sleep to determine if there is an association with overall quarterly academic performance, measured by average quarter GPA.

### Dataset #4: CMU Sleep Study: Sleep Duration, Consistency, and GPA

This dataset comes from a published academic study tracking the sleep behavior and academic performance of 634 first-year college students across three U.S. universities: Carnegie Mellon University, University of Washington, and University of Notre Dame. Sleep was recorded using Fitbit wearable devices over approximately one month during the spring academic term, and GPA records were obtained directly from university registrars, making the academic performance measure more reliable than self-reported alternatives.

The most important metrics in this dataset are as follows. `TotalSleepTime` measures average nightly sleep duration in minutes — healthy sleep for young adults falls between 420 and 540 minutes (7–9 hours), and values below 360 minutes (6 hours) have been associated with significant cognitive impairment and lower academic performance. `bedtime_mssd` is the mean successive squared difference of bedtime in minutes squared, measuring night-to-night variability in sleep timing; a higher value indicates more irregular sleep schedules, which is directly relevant to our consistency hypothesis. `midpoint_sleep` is the average midpoint of the sleep episode measured in minutes after 11PM — for example, a value of 240 corresponds to a sleep midpoint of 3AM, indicating a late sleep schedule that overlaps with our 11PM–6AM window of interest. `daytime_sleep` captures average minutes of sleep outside the main nightly episode; elevated values may suggest chronic sleep deprivation being compensated through napping. `term_gpa` and `cum_gpa` both measure academic performance on a 4.0 scale, with `term_gpa` reflecting the specific term during which sleep was tracked and `cum_gpa` reflecting prior academic history.

A major concern with this dataset is that it contains no smartphone screen time data whatsoever, meaning it can only address the sleep → GPA portion of our research question and not the screen time → GPA portion directly. Additionally, the sample is restricted to first-year students, who may behave differently from upperclassmen in terms of sleep habits and academic adjustment. The study also only tracked sleep for roughly one month rather than the full academic quarter, which may not capture the full range of sleep behavior over a term. Finally, participants were drawn from only three universities, limiting generalizability to broader college populations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import os

# ── 1. Load raw data ──────────────────────────────────────────────────────────
cmu_raw = pd.read_csv('data/00-raw/cmu-sleep.csv')

print("Shape:", cmu_raw.shape)
print("\nColumn names:", cmu_raw.columns.tolist())
print("\nData types:\n", cmu_raw.dtypes)
cmu_raw.head()
```

```python
# ── 2. Demonstrate tidiness ───────────────────────────────────────────────────
# Tidy data: each row = one student (one observation), each column = one variable
# This dataset is already in wide format with one row per student — already tidy.
print("Number of duplicate rows:", cmu_raw.duplicated().sum())
print("\nEach row represents one student — dataset is tidy.")
```

```python
# ── 3. Demonstrate size ───────────────────────────────────────────────────────
print(f"Number of observations (rows): {cmu_raw.shape[0]}")
print(f"Number of variables (columns): {cmu_raw.shape[1]}")
```

```python
# ── 4. Missingness analysis ───────────────────────────────────────────────────
missing = cmu_raw.isnull().sum()
missing_pct = (missing / len(cmu_raw) * 100).round(2)
missing_df = pd.DataFrame({
    'missing_count': missing, 
    'missing_pct': missing_pct
})
print("Missing data summary:")
print(missing_df[missing_df['missing_count'] > 0])

# Visualize missingness
if missing_df['missing_count'].sum() > 0:
    plt.figure(figsize=(10, 4))
    missing_pct[missing_pct > 0].sort_values().plot(kind='barh', color='steelblue')
    plt.xlabel('% Missing')
    plt.title('Missing Data by Column — CMU Sleep')
    plt.tight_layout()
    plt.show()
else:
    print("No missing values found.")
```

```python
# ── 5. Outlier detection ──────────────────────────────────────────────────────
key_vars = ['TotalSleepTime', 'bedtime_mssd', 'midpoint_sleep', 
            'daytime_sleep', 'cum_gpa', 'term_gpa']
key_vars = [v for v in key_vars if v in cmu_raw.columns]

# Boxplots to visually identify outliers
fig, axes = plt.subplots(1, len(key_vars), figsize=(18, 4))
for ax, col in zip(axes, key_vars):
    cmu_raw[col].dropna().plot(kind='box', ax=ax, title=col)
plt.suptitle('Boxplots of Key Variables — CMU Sleep', y=1.02)
plt.tight_layout()
plt.show()

# Flag suspicious entries
# TotalSleepTime: < 60 min likely device error; > 720 min (12 hrs) implausible
if 'TotalSleepTime' in cmu_raw.columns:
    suspicious_sleep = cmu_raw[
        (cmu_raw['TotalSleepTime'] < 60) | (cmu_raw['TotalSleepTime'] > 720)
    ]
    print(f"Suspicious TotalSleepTime entries (< 60 or > 720 min): {len(suspicious_sleep)}")

# GPA: flag values outside valid 0.0–4.0 range
for gpa_col in ['cum_gpa', 'term_gpa']:
    if gpa_col in cmu_raw.columns:
        bad_gpa = cmu_raw[(cmu_raw[gpa_col] < 0) | (cmu_raw[gpa_col] > 4.0)]
        print(f"Out-of-range {gpa_col} entries: {len(bad_gpa)}")
```

```python
# ── 6. Clean the data ─────────────────────────────────────────────────────────
cmu_clean = cmu_raw.copy()

# Drop rows missing any key analysis variables
# Justification: all three variables (sleep duration, sleep consistency, GPA) 
# are required for our analysis — a row missing any of them is unusable.
cmu_clean = cmu_clean.dropna(subset=key_vars, how='any')

# Remove physiologically implausible sleep values
if 'TotalSleepTime' in cmu_clean.columns:
    cmu_clean = cmu_clean[
        (cmu_clean['TotalSleepTime'] >= 60) & (cmu_clean['TotalSleepTime'] <= 720)
    ]

# Remove out-of-range GPA values
for gpa_col in ['cum_gpa', 'term_gpa']:
    if gpa_col in cmu_clean.columns:
        cmu_clean = cmu_clean[
            (cmu_clean[gpa_col] >= 0) & (cmu_clean[gpa_col] <= 4.0)
        ]

print(f"Shape after cleaning: {cmu_clean.shape}")
print(f"Rows dropped: {len(cmu_raw) - len(cmu_clean)}")
```

```python
# ── Rename columns for clarity ─────────────────────────────────────────────
cmu_clean = cmu_clean.rename(columns={
    'subject_id': 'student_id',
    'study': 'study_id',
    'cohort': 'cohort_group',
    'demo_race': 'race',
    'demo_gender': 'gender',
    'demo_firstgen': 'is_first_gen_student',
    'bedtime_mssd': 'bedtime_night_to_night_variability',
    'TotalSleepTime': 'avg_nightly_sleep_minutes',
    'midpoint_sleep': 'avg_sleep_midpoint_minutes_after_11pm',
    'frac_nights_with_data': 'fraction_of_nights_with_recorded_data',
    'daytime_sleep': 'avg_daytime_nap_minutes',
    'cum_gpa': 'cumulative_gpa',
    'term_gpa': 'term_gpa',
    'term_units': 'units_enrolled',
    'Zterm_units_ZofZ': 'units_enrolled_standardized'
})

print("Renamed columns:", cmu_clean.columns.tolist())
```

```python
# ── 7. Save to processed ──────────────────────────────────────────────────────
os.makedirs('data/02-processed', exist_ok=True)
cmu_clean.to_csv('data/02-processed/cmu_sleep_clean.csv', index=False)
print("Saved to data/02-processed/cmu_sleep_clean.csv")
```

#### CMU Sleep Dataset: Multivariable Relational Analysis

We now examine the relationships between our key predictors (sleep duration and sleep consistency) and our outcome variable (term GPA). We use binned permutation tests to assess statistical significance and simple linear regression to quantify the direction and strength of each relationship.

```python
# ── Sleep duration vs GPA — binned permutation test ─────────────────
sleep_col = 'avg_nightly_sleep_minutes'
cmu_clean['sleep_group'] = pd.cut(cmu_clean[sleep_col], 
                                   bins=3, 
                                   labels=['Low Sleep', 'Medium Sleep', 'High Sleep'])

low_gpa = cmu_clean[cmu_clean['sleep_group'] == 'Low Sleep']['term_gpa'].dropna()
high_gpa = cmu_clean[cmu_clean['sleep_group'] == 'High Sleep']['term_gpa'].dropna()
observed_diff_sleep = high_gpa.mean() - low_gpa.mean()
print(f"Observed GPA difference (High - Low sleep): {observed_diff_sleep:.4f}")

n_permutations = 1000
combined = np.concatenate([low_gpa, high_gpa])
perm_diffs = []

for _ in range(n_permutations):
    shuffled = combined.copy()
    np.random.shuffle(shuffled)
    perm_diffs.append(shuffled[len(low_gpa):].mean() - shuffled[:len(low_gpa)].mean())

p_value_sleep = np.mean(np.abs(perm_diffs) >= np.abs(observed_diff_sleep))
print(f"Permutation test p-value (sleep vs GPA): {p_value_sleep:.4f}")

plt.figure(figsize=(8, 5))
plt.hist(perm_diffs, bins=40, color='steelblue', edgecolor='white', alpha=0.8)
plt.axvline(observed_diff_sleep, color='red', linestyle='--', 
            label=f'Observed diff: {observed_diff_sleep:.3f}')
plt.xlabel('Permuted GPA Difference')
plt.ylabel('Count')
plt.title('Permutation Test: Sleep Duration vs GPA — CMU Sleep')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ── Boxplot of GPA by sleep group ────────────────────────────────────
plt.figure(figsize=(8, 5))
cmu_clean.boxplot(column='term_gpa', by='sleep_group')
plt.suptitle('')
plt.title('Term GPA by Sleep Group — CMU Sleep Dataset')
plt.xlabel('Sleep Group')
plt.ylabel('Term GPA')
plt.tight_layout()
plt.show()
```

```python
# ── Sleep consistency vs GPA — binned permutation test ───────────────
consist_col = 'bedtime_night_to_night_variability'
cmu_clean['consistency_group'] = pd.cut(cmu_clean[consist_col],
                                         bins=3,
                                         labels=['Consistent', 'Moderate', 'Inconsistent'])

consistent_gpa = cmu_clean[cmu_clean['consistency_group'] == 'Consistent']['term_gpa'].dropna()
inconsistent_gpa = cmu_clean[cmu_clean['consistency_group'] == 'Inconsistent']['term_gpa'].dropna()
observed_diff_consist = consistent_gpa.mean() - inconsistent_gpa.mean()
print(f"Observed GPA difference (Consistent - Inconsistent): {observed_diff_consist:.4f}")

combined2 = np.concatenate([consistent_gpa, inconsistent_gpa])
perm_diffs2 = []

for _ in range(n_permutations):
    shuffled2 = combined2.copy()
    np.random.shuffle(shuffled2)
    perm_diffs2.append(shuffled2[:len(consistent_gpa)].mean() - shuffled2[len(consistent_gpa):].mean())

p_value_consist = np.mean(np.abs(perm_diffs2) >= np.abs(observed_diff_consist))
print(f"Permutation test p-value (consistency vs GPA): {p_value_consist:.4f}")

plt.figure(figsize=(8, 5))
plt.hist(perm_diffs2, bins=40, color='orange', edgecolor='white', alpha=0.8)
plt.axvline(observed_diff_consist, color='red', linestyle='--',
            label=f'Observed diff: {observed_diff_consist:.3f}')
plt.xlabel('Permuted GPA Difference')
plt.ylabel('Count')
plt.title('Permutation Test: Sleep Consistency vs GPA — CMU Sleep')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ── Simple linear regression ─────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, col, label in zip(axes,
                           ['avg_nightly_sleep_minutes', 
                            'bedtime_night_to_night_variability'],
                           ['Sleep Duration (min)', 
                            'Sleep Consistency (MSSD)']):
    valid = cmu_clean[[col, 'term_gpa']].dropna()
    x = valid[col]
    y = valid['term_gpa']
    
    slope, intercept, r, p, se = stats.linregress(x, y)
    
    ax.scatter(x, y, alpha=0.4, color='steelblue')
    ax.plot(x, slope * x + intercept, color='red', linewidth=2,
            label=f'r={r:.3f}, p={p:.4f}')
    ax.set_xlabel(label)
    ax.set_ylabel('Term GPA')
    ax.set_title(f'{label} vs Term GPA')
    ax.legend()

plt.suptitle('Simple Linear Regression — CMU Sleep Dataset', fontsize=14)
plt.tight_layout()
plt.show()
```

#### CMU Sleep Dataset: Confounding Variable Analysis

Several demographic and academic variables in this dataset could confound the sleep-GPA relationship. We examine whether gender, race, first-generation student status, and units enrolled are correlated with our key predictors and outcome, and use multiple linear regression to test whether the sleep-GPA association holds after controlling for these factors.

```python
# ── Correlations between confounders and key variables ───────────────
confounders = ['gender', 'race', 'is_first_gen_student', 'units_enrolled']
confounders = [c for c in confounders if c in cmu_clean.columns]

# Convert confounders to numeric, coercing non-numeric values to NaN
for col in confounders:
    cmu_clean[col] = pd.to_numeric(cmu_clean[col], errors='coerce')

vars_of_interest = ['avg_nightly_sleep_minutes',
                    'bedtime_night_to_night_variability',
                    'term_gpa']

print("=== Correlations: Confounders vs Key Variables ===")
for conf in confounders:
    print(f"\n--- {conf} ---")
    for var in vars_of_interest:
        try:
            valid = cmu_clean[[conf, var]].dropna()
            r, p = stats.pearsonr(valid[conf], valid[var])
            print(f"  vs {var}: r={r:.3f}, p={p:.4f}")
        except Exception as e:
            print(f"  vs {var}: could not compute ({e})")
```

```python
# ── Multiple linear regression controlling for confounders ────────────
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

reg_vars = ['avg_nightly_sleep_minutes', 
            'bedtime_night_to_night_variability'] + confounders
reg_data = cmu_clean[reg_vars + ['term_gpa']].copy()

# Convert all columns to numeric, coercing any non-numeric values to NaN
for col in reg_data.columns:
    reg_data[col] = pd.to_numeric(reg_data[col], errors='coerce')

reg_data = reg_data.dropna()
print(f"Rows used in regression: {len(reg_data)}")

X = reg_data[reg_vars]
y = reg_data['term_gpa']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = LinearRegression()
model.fit(X_scaled, y)

print("\n=== Multiple Linear Regression: Predicting Term GPA ===")
print(f"R² score: {model.score(X_scaled, y):.4f}")
print("\nStandardized Coefficients:")
for name, coef in zip(reg_vars, model.coef_):
    print(f"  {name}: {coef:.4f}")
```

```python
# ── Subgroup analysis by gender ──────────────────────────────────────
if 'gender' in cmu_clean.columns:
    # Convert gender to numeric, dropping any non-numeric values
    cmu_clean['gender'] = pd.to_numeric(cmu_clean['gender'], errors='coerce')
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    for ax, col, xlabel in zip(axes,
                                ['avg_nightly_sleep_minutes',
                                 'bedtime_night_to_night_variability'],
                                ['Sleep Duration (min)',
                                 'Sleep Consistency (MSSD)']):
        for gender_val in cmu_clean['gender'].dropna().unique():
            subset = cmu_clean[cmu_clean['gender'] == gender_val].dropna(
                subset=[col, 'term_gpa'])
            ax.scatter(subset[col], subset['term_gpa'],
                      alpha=0.4, label=f'Gender {int(gender_val)}')
        ax.set_xlabel(xlabel)
        ax.set_ylabel('Term GPA')
        ax.set_title(f'{xlabel} vs GPA by Gender')
        ax.legend()

    plt.suptitle('Subgroup Analysis by Gender — CMU Sleep Dataset', fontsize=14)
    plt.tight_layout()
    plt.show()
```

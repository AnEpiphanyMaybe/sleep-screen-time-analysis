## Research Question

Does consistency and length of smartphone usage, between 11PM to 6AM, of California university students have an association with overall quarterly academic performance, measured by average quarter GPA?

## Background and Prior Work


<p>Smartphones have become an almost inseparable part of college student’s lives, and consequently, the habit of using them late into the night has become increasingly common. While there is lots of documentation on how poor sleep affects academic performance, less attention has been given to the role late night smartphone usage (between 11PM - 6AM) plays in that process. College students need to balance between late-night entertainment, messaging, and media scrolling with the sleep time they need to reinforce what they learned and to perform well academically. The 11PM threshold is chosen because it represents the boundary after which blue light exposure from screens most aggressively suppresses the production of melatonin and delays sleep, and it aligns with the recommended bedtime range for young adults seeking 7-9 hours of sleep to have healthy cognitive function. <a name="cite_ref-7"></a><a href="#cite_note-7"><sup>7</sup></a>

Instead of directly measuring sleep, this project uses late-time smartphone usage as a behavioral proxy, which is more objective and more actionable than self-reported sleep quality, questioning whether the length of time and consistency of phone use during these hours is in correlation to quarterly GPA. <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Learning about this relationship can help students and institutions identify a solid, measurable behavior, one that students can actually see their own screen time reports, that might be unknowingly undermining their academic performance.

It is well established that the general usage of smartphones can cause physical and mental discomfort. In addition, increasing the amount of cellphone usage before bedtime has been linked to making the user more susceptible to psychological unease and sleep disorders. <a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) Electronic devices such as smartphones emit blue light, which is high in energy and short in wavelength. Long exposure to this affects sleep by inhibiting melatonin production, creating circadian disruptions, leading to a deterioration of both sleep quality and duration. <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) With the prevalence of electronic devices for adolescents, there have been many studies on whether smart phone usage has an impact on academic performance. Many studies highlighted a negative relationship between bedtime phone usage and academic performance, with many students with less hours of sleep due to phone usage reporting a higher tiredness associated with low productivity. <a name="cite_ref-2"></a>[<sup>2,</sup>](#cite_note-2)<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) It’s also observed that more bedtime smart phone usage leads to nomophobia development, or an anxiety of being without a mobile device, which resulted in lower academic scores and ranking. [<sup>5</sup>](#cite_note-5) The impact of late-night phone use has also been observed in younger students, where smartphone usage was found to negatively affect learning effectiveness and academic engagement. <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) Many of these studies used self reported data, which is limited by bias and validity. They also ask students to report how they feel rather than use an arbitrary measurement, which can also increase the likelihood of bias. [<sup>3</sup>](#cite_note-3) We want to use screentime, frequency of late night phone. </p>

## Hypothesis


We predict a negative correlation between quarterly GPA and the length of smartphone usage between 11PM and 6AM. We also expect that greater consistency in late-night smartphone usage is positively correlated with GPA. This is because we think longer late-night screen time likely reduces sleep time and sleep quality, although consistency of late night usage patterns may make the bodies and schedules of individuals get used to their later schedules, making these negative effects less impactful.

## Data

<p>Dataset 1: The StudentLife dataset is available on Kaggle at https://www.kaggle.com/datasets/dartweichen/student-life and contains smartphone sensing data collected from Dartmouth students. The dataset is public although the sample is mid-sized and restricted to a single non-California university. Important variables from the dataset that we might use include timestamped phone activity usage as well as GPA.

Dataset 2: The Add Health dataset is available through UNC Dataverse at https://dataverse.unc.edu/dataverse/addhealth and is a study tracking U.S. adolescents and young adults. The dataset lacks timestamped usage data requiring approximation. Important variables that we might use include academic transcript data and behavioral measures such as self-reported screen use and sleep duration. This could be combined to create proxy measurements for nighttime phone usage. Users are required to sign a Dataset Terms form and agree to the Add Health Dataverse terms of use in order to access the dataset.

Dataset 3: https://www.kaggle.com/datasets/rishisukumar/student-screen-time-vs-cgpa-analysis-2026 Public dataset. Based on 550 undergraduates across India. It has no specific data on night time screen time but has daily screen time data, a breakdown of time spent on different types of screen usage (social media, studying, gaming), discipline during the day, and comparison of GPA across two semesters to track academic performance over a longer period of time.

Dataset 4: https://www.kaggle.com/datasets/mexwell/cmu-sleep Public dataset. It has data of 634 student participants from three universities (CMU, UW and ND) across US, focused on relation between sleep and GPA but doesn’t include screen time info, and needs to be joined with other datasets. Important variables include total sleep time average, midpoint of bedtime and wake time, bedtime mssd (variation of bedtime over consecutive nights), units taken in the term, cumulative gpa and term gpa.

Dataset 5: https://www.kaggle.com/datasets/hammadansari7/gen-z-mental-wellness-and-digital-lifestyle-patterns Public dataset. This dataset is synthetic but contains some useful information. It contains multiple indicators of digital behavior, such as average daily social media usage, total daily screen time, late-night phone usage frequency, daily gaming hour, and primary digital content category preference. Along with multiple indicator variables describing the mental health status of the individuals, such as sleep quality, anxiety level, and motivation level, and lifestyle indicators such as exercise frequency per week, and study work hours per day, the dataset connects screen usage with the overall well-being of individuals in the age range of 18-26.</p>
